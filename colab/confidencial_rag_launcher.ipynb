# Confidencial RAG Colab Launcher
Run cells top to bottom. The shared Gradio URL is internet-accessible; use temporary credentials only.


```python
import subprocess
import sys
from pathlib import Path
REPO_URL = 'https://github.com/leodman/Confidencial-RAG.git'
REPO_DIR = Path('/content/Confidencial-RAG')
if REPO_DIR.exists():
    subprocess.check_call(['git','-C',str(REPO_DIR),'fetch','origin','main'])
    subprocess.check_call(['git','-C',str(REPO_DIR),'checkout','main'])
    subprocess.check_call(['git','-C',str(REPO_DIR),'pull','--ff-only','origin','main'])
else:
    subprocess.check_call(['git','clone',REPO_URL,str(REPO_DIR)])
    subprocess.check_call(['git','-C',str(REPO_DIR),'checkout','main'])
print('Repository ready:', REPO_DIR)
```

```python
import subprocess
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', '-e', str(REPO_DIR) + '[ui,rag,documents]'])
SRC = str(REPO_DIR / 'src')
if SRC not in sys.path:
    sys.path.insert(0, SRC)
import confidencial_rag
print('confidencial_rag', confidencial_rag.__version__)
try:
    import gradio
    import numpy
    print('gradio', gradio.__version__, 'numpy', numpy.__version__)
except Exception as exc:
    print('Diagnostic version check warning:', type(exc).__name__)
```

```python
from getpass import getpass
from confidencial_rag.ui.colab import launch
consent = input('Create a temporary internet-accessible Gradio shared URL? Type YES to continue: ')
if consent != 'YES':
    raise SystemExit('Launch cancelled by user.')
username = input('Temporary Gradio username: ').strip()
password = getpass('Temporary Gradio password (not saved): ')
if not username or not password:
    raise SystemExit('Username and password are required for authenticated sharing.')
print('WARNING: The Gradio shared URL is internet-accessible. Authentication reduces exposure but is not private infrastructure.')
launch(share=True, username=username, password=password)
```
