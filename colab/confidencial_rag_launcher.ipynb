# Confidencial RAG launcher
This thin launcher installs and starts the application shell. Use only synthetic data in this milestone.

```python
from pathlib import Path
import subprocess

repo = Path('/content/Confidencial-RAG')
if (repo / '.git').exists():
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', 'https://github.com/leodman/Confidencial-RAG.git', str(repo)], check=True)
subprocess.run(['python', '-m', 'pip', 'install', '-e', f'{repo}[ui]'], check=True)
```

```python
from getpass import getpass
from confidencial_rag.ui import launch

enable_share = input('Enable a Gradio shared URL? [y/N]: ').strip().lower() == 'y'
username = password = None
if enable_share:
    username = input('Shared-URL username: ').strip()
    password = getpass('Shared-URL password: ')
    if not username or not password:
        raise ValueError('Sharing requires a non-empty username and password.')

launch(share=enable_share, username=username, password=password)
```
