# Confidencial RAG launcher
This thin launcher installs the application shell. Google Colab cannot expose a localhost-only Gradio page to your browser, so opening the interface requires an explicitly enabled temporary Gradio shared URL. Treat that URL as internet-accessible and use only synthetic data in this milestone.

```python
from pathlib import Path
import subprocess
import sys

repo = Path('/content/Confidencial-RAG')
if (repo / '.git').exists():
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', 'https://github.com/leodman/Confidencial-RAG.git', str(repo)], check=True)
subprocess.run(
    [
        sys.executable,
        '-m',
        'pip',
        'install',
        '-e',
        f'{repo}[ui]',
    ],
    check=True,
)

import confidencial_rag  # noqa: E402

print('Confidencial RAG installed from:', confidencial_rag.__file__)
```

```python
from getpass import getpass
from confidencial_rag.ui import launch_from_colab

enable_share = input('Create a temporary Gradio shared URL? [y/N]: ').strip().lower() == 'y'
username = password = None
if enable_share:
    username = input('Shared-URL username: ').strip()
    password = getpass('Shared-URL password: ')

message = launch_from_colab(
    enable_share,
    username=username,
    password=password,
)
print(message)
```
